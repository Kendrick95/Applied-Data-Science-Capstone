```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

# Example usage:
launch_site_lat = 28.5623
launch_site_lon = -80.5774
coastline_lat = 28.5620  # replace with actual coordinate
coastline_lon = -80.5850  # replace with actual coordinate

distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f"Distance from launch site to coastline: {distance_coastline:.2f} km")
```

```python
# Example coordinates (replace with your actual coordinates)
launch_site_coord = [28.5623, -80.5774]
closest_point_coord = [28.5650, -80.5800]  # e.g., closest city, railway, or highway

# Function to calculate distance (Haversine formula)
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

# Calculate distance between launch site and closest point
distance = calculate_distance(
    launch_site_coord[0], launch_site_coord[1],
    closest_point_coord[0], closest_point_coord[1]
)

# Initialize the map centered near the launch site
site_map = folium.Map(location=launch_site_coord, zoom_start=10)

# Add a marker for the launch site
folium.Marker(
    location=launch_site_coord,
    popup='Launch Site',
    icon=folium.Icon(color='blue', icon='rocket')
).add_to(site_map)

# Add a marker for the closest point with distance label
distance_marker = folium.Marker(
    location=closest_point_coord,
    icon=DivIcon(
        icon_size=(150, 36),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12px; color:#d35400;"><b>{:10.2f} KM</b></div>'.format(distance),
    )
)
distance_marker.add_to(site_map)

# Draw a line between the launch site and the closest point
folium.PolyLine(locations=[launch_site_coord, closest_point_coord], color='green', weight=2).add_to(site_map)

# Display the map
site_map
```
